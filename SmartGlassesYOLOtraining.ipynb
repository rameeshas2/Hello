### **STT_Engine.py**

```python
%%writefile translation/stt_engine.py
import whisper

class STTEngine:
    def __init__(self):
        self.model = whisper.load_model("base")

    def transcribe(self, audio_path):
        result = self.model.transcribe(audio_path, language='ur')
        return result['text']
```

## **TTS_Engigne.py**

```python
%%writefile translation/tts_engine.py
import pyttsx3

class TTSEngine:
    def __init__(self):
        self.engine = pyttsx3.init()

    def speak(self, text):
        self.engine.say(text)
        self.engine.runAndWait()
```

### **Main_pipe.py**

```python
%%writefile translation/main_pipe.py
from stt_engine import STTEngine
from tts_engine import TTSEngine
from deep_translator import GoogleTranslator

def run_translation(audio_file):
    stt = STTEngine()
    tts = TTSEngine()

    # 1. Transcribe Urdu
    urdu_text = stt.transcribe(audio_file)
    print(f"Urdu: {urdu_text}")

    # 2. Translate to English
    english_text = GoogleTranslator(source='ur', target='en').translate(urdu_text)
    print(f"English: {english_text}")

    # 3. Speak
    tts.speak(english_text)

if __name__ == "__main__":
    # Ensure you upload an audio file named 'input_audio.wav' to the 'translation' folder
    run_translation('translation/input_audio.wav')
```
